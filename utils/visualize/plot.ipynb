```python
import matplotlib.pyplot as plt
import numpy as np

# 设置编辑强度范围
gamma = np.linspace(0, 1, 100)

# 正确语义方向的图像轨迹（例如微笑）
image_correct_x = gamma
image_correct_y = np.sin(gamma * np.pi / 2) * 0.3

# entangled 的图像方向轨迹（例如微笑 + 姿态变化）
image_entangled_y = -np.cos(gamma * np.pi) * 0.2

# 文本空间中起始和目标点（离散的）
text_point_start = (0.0, 0.8)
text_point_end = (1.0, 0.9)

# 图像方向起点与两个路径终点
img_point_start = (0.0, 0.0)
img_point_correct = (1.0, image_correct_y[-1])
img_point_wrong = (1.0, image_entangled_y[-1])

# 绘图
plt.figure(figsize=(8, 6))

# 正确方向轨迹
plt.plot(image_correct_x, image_correct_y, color='blue', label='Correct image edit direction (e.g., smiling)')
# 错误方向轨迹（entangled）
plt.plot(image_correct_x, image_entangled_y, color='purple', linestyle='--', label='Entangled edit direction')

# 文本方向向量
plt.scatter(*text_point_start, color='red')
plt.text(*text_point_start, '"a person"', fontsize=9, color='red', ha='right')
plt.scatter(*text_point_end, color='red')
plt.text(*text_point_end, '"a smiling person"', fontsize=9, color='red', ha='left')
plt.arrow(*text_point_start,
          text_point_end[0] - text_point_start[0],
          text_point_end[1] - text_point_start[1],
          color='red', width=0.003, head_width=0.02, length_includes_head=True,
          label='Text direction in CLIP space')

# 图像方向箭头：正确方向
plt.arrow(*img_point_start,
          img_point_correct[0] - img_point_start[0],
          img_point_correct[1] - img_point_start[1],
          color='gray', linestyle='-', width=0.0015, head_width=0.02,
          label='Image direction (aligned)')

# 图像方向箭头：错误方向
plt.arrow(*img_point_start,
          img_point_wrong[0] - img_point_start[0],
          img_point_wrong[1] - img_point_start[1],
          color='black', linestyle='--', width=0.0015, head_width=0.02,
          label='Image direction (entangled)')

# 说明文字
# plt.text(0.25, 0.45,
#          "Multiple image directions may align\nwith the same text prompt\nCLIP loss cannot distinguish them",
#          fontsize=10, color='black')

# 图像配置
plt.xlabel("Edit trajectory in activation space (Δh direction)")
plt.ylabel("CLIP embedding projection")
plt.title("Directional CLIP Loss Allows Entangled Image Directions")
plt.grid(True)
plt.legend()
plt.tight_layout()
plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np
import matplotlib
# 编辑强度范围
matplotlib.rcParams['font.family'] = 'Times New Roman'



gamma = np.linspace(0, 1, 100)

# 正确语义方向的图像轨迹
image_correct_x = gamma
image_correct_y = np.sin(gamma * np.pi / 2) * 0.3

# entangled 的图像方向轨迹
image_entangled_y = -np.cos(gamma * np.pi) * 0.2

# 文本空间中起始和目标点
text_point_start = (0.0, 0.3)
text_point_end = (1.0, 0.5)

# 图像方向起点与终点
img_point_start = (0.0, 0.0)
img_point_correct = (1.0, image_correct_y[-1])
img_point_wrong = (1.0, image_entangled_y[-1])

plt.figure(figsize=(8, 6))

# 正确图像轨迹
plt.plot(image_correct_x, image_correct_y, color='blue', label='Correct image edit direction (e.g., smiling)')
# 错误图像轨迹（entangled）
plt.plot(image_correct_x, image_entangled_y, color='purple', linestyle='--', label='Entangled edit direction')

# 文本方向向量
plt.scatter(*text_point_start, color='red')
plt.scatter(*text_point_end, color='red')
plt.arrow(*text_point_start,
          text_point_end[0] - text_point_start[0],
          text_point_end[1] - text_point_start[1],
          color='red', width=0.003, head_width=0.02, length_includes_head=True,
          label='Text direction in CLIP space')

# 文本标注位置优化
plt.text(text_point_start[0] - 0.05, text_point_start[1] - 0.05, '"a person"', fontsize=9, color='red')
plt.text(text_point_end[0] - 0.11 , text_point_end[1] - 0.05 , '"a smiling person"', fontsize=9, color='red')

# 图像方向箭头：正确方向
plt.arrow(*img_point_start,
          img_point_correct[0] - img_point_start[0],
          img_point_correct[1] - img_point_start[1],
          color='gray', linestyle='-', width=0.0015, head_width=0.02,
          label='Image direction (aligned)')

# 图像方向箭头：错误方向
plt.arrow(*img_point_start,
          img_point_wrong[0] - img_point_start[0],
          img_point_wrong[1] - img_point_start[1],
          color='black', linestyle='--', width=0.0015, head_width=0.02,
          label='Image direction (entangled)')

# 说明文字放置更自然的位置
# plt.text(0.6, 0.05,
#          "Multiple image directions may align\nwith the same text prompt\nCLIP loss cannot distinguish them",
#          fontsize=10, color='black')

# 图像外观设置
plt.xlabel("Edit trajectory in activation space (Δh direction)")
plt.ylabel("CLIP embedding projection")
plt.title("Directional CLIP Loss Allows Entangled Image Directions")
plt.grid(True)
plt.legend(loc='lower right')  # 图例移到底部右侧
plt.tight_layout()

plt.savefig("clip_loss_entangled.pdf", format='pdf', bbox_inches='tight')
plt.show()
```
